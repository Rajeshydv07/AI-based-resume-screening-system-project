```python
import os
import sys
import pandas as pd

# Load saved baseline results
baseline_path = '../models/baseline_results.csv'
if os.path.exists(baseline_path):
    results_df = pd.read_csv(baseline_path)
    print('=== 5-FOLD CROSS VALIDATION & TEST EVALUATION ===')
    display(results_df)
else:
    sys.path.append(os.path.join('..', 'src'))
    from model_trainer import train_baseline_models
    results_df, _ = train_baseline_models('../data', '../models')
    display(results_df)
```

### Performance Comparison Chart
Compare baseline models across Accuracy and F1 Score.

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(9, 4))
results_df.plot(x='Model', y=['Test Accuracy', 'Test F1'], kind='bar', ax=ax, colormap='Blues_r')
plt.title('Baseline Models Comparison on Test Data')
plt.ylabel('Score')
plt.xticks(rotation=15)
plt.legend(['Test Accuracy', 'Test F1'])
plt.ylim(0.5, 1.0)
plt.tight_layout()
plt.show()
```
