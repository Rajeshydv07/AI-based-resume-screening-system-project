# Week 3: Data Cleaning & Text Normalization
Final-Year B.Tech IT Project | AI-Based Resume Screening System

```python
import os
import sys
import pandas as pd
import numpy as np

sys.path.append(os.path.join('..', 'src'))
from data_cleaner import remove_duplicates, fill_missing, cap_outliers, clean_text

df_raw = pd.read_csv('../data/raw_resumes.csv')
print('Raw dataset shape:', df_raw.shape)
```

### 1. Duplicate Removal
Identify and drop duplicate candidate profiles.

```python
df_dedup, num_dups = remove_duplicates(df_raw)
print(f'Duplicates dropped: {num_dups}')
print(f'Dataset shape after deduplication: {df_dedup.shape}')
```

### 2. Imputation of Missing Values
Fill missing numerical values with median and categorical with mode.

```python
df_imputed = fill_missing(df_dedup)
print('Remaining missing values:')
print(df_imputed.isnull().sum())
```

### 3. Outlier Capping (Winsorization)
Cap extreme values to valid numerical thresholds.

```python
df_capped, outliers_dict = cap_outliers(df_imputed)
print('Outliers capped summary:', outliers_dict)
print('Experience range after capping:', df_capped['Experience_Years'].min(), 'to', df_capped['Experience_Years'].max())
print('CGPA range after capping:', df_capped['CGPA'].min(), 'to', df_capped['CGPA'].max())
```

### 4. Text Normalization
Lowercase, strip emails/URLs, remove stopwords and lemmatize resume text.

```python
sample_text = df_capped['Resume_Text'].iloc[0]
print('Original text preview:')
print(sample_text[:120])
print('\nCleaned text preview:')
print(clean_text(sample_text)[:120])

# Clean all text
df_capped['Cleaned_Resume_Text'] = df_capped['Resume_Text'].apply(clean_text)
print('Cleaned dataset ready! Shape:', df_capped.shape)
```
